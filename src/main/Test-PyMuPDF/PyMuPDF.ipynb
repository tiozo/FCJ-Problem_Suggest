```python
doc = fitz.open("de-thi-thu-tn-thpt-2025-mon-toan-lan-2-truong-thpt-huong-hoa-quang-tri.pdf")

for i in range(doc.page_count):
    page = doc[i]
    img = page.get_pixmap()
    img.save(f"page-{i}.png")
```

```python
!pip install pdfplumber
import pdfplumber
```

```python
with pdfplumber.open("de-thi-thu-tn-thpt-2025-mon-toan-lan-2-truong-thpt-huong-hoa-quang-tri.pdf") as pdf:
    for i in range(len(pdf.pages)):
        page = pdf.pages[i]
        page.to_image().save(f"page-{i}.png")
```

```python
!pip install PyPDF2
import PyPDF2
```

```python
reader = PyPDF2.PdfReader("de-thi-thu-tn-thpt-2025-mon-toan-lan-2-truong-thpt-huong-hoa-quang-tri.pdf")

for i in range(len(reader.pages)):
    page = reader.pages[i]
    count = 0
    print(i)
    for image_file_object in page.images:
        with open(str(count) + image_file_object.name, "wb") as fp:
            fp.write(image_file_object.data)
            count += 1
```
